````python
prompts = f"""
    You are expert data extraction and structuring agent, return the following in JSON structured format.
    
    Extract the following fields from the data:
    - Patient name (from Childs NameID)
    - Date of birth (from Childs Date of Birth)
    - Assessment date (from Todays Date)
    - ALL observations (from Score1 through Score63)

    IMPORTANT INSTRUCTIONS:
    1. Extract ALL scores from Score1 to Score63
    2. Each score should be included as an observation
    3. Include the observation number (1-63) and the full description
    4. Do not skip any scores
    5. Return only the JSON response

    JSON RESPONSE FORMAT:
    {{
        "Patient name": "**REPLACE THIS WITH NAME OF THE PATIENT**",
        "Date of birth": "**REPLACE WITH PATIENT DATE OF BIRTH**",
        "Assessment date": "**REPLACE WITH ASSESSMENT DATE**",
        "Observations": [
            {{
                "Observation no": "1",
                "Observation": "stand without holding on to anything",
                "Value": "2"
            }},
            {{
                "Observation no": "2",
                "Observation": "walk 10-20 steps by himself/herself",
                "Value": "2"
            }},
            ... CONTINUE FOR ALL 63 SCORES ...
        ]
    }}

    RAW DATA:
    {json_response}
"""
````

````python
response = client.invoke(prompts)
response
````

````python
ai_response_data = response.content.replace("```json", "").replace("```", "")
ai_response_json = json.loads(ai_response_data)
print(json.dumps(ai_response_json, indent=4))
````
