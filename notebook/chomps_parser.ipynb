```python


# Load CSV
df: pd.DataFrame | None = None
for file in FILENAME:
    if file.endswith(".json"): continue
    print(file)
    if not df:
        df = pd.read_csv(file, header=None)
        
    df = df + pd.read_csv(file, header=None)
    # print(pd.read_csv(file, header=None))
```

```python

# Helper function to identify rows starting with item numbers - FIXED
def is_item_row(cell):
    if isinstance(cell, str):
        return bool(re.match(r"^\s*'?[\d]+\.", cell.strip()))
    return False

# Filter relevant rows that start with item numbers
item_rows = df[df[0].apply(is_item_row)].copy()

# Clean and rename columns
item_rows.columns = ['item_description', 'yes', 'sometimes', 'not_yet', 'score', 'extra']
item_rows.drop(columns=['extra'], inplace=True)

# Remove leading/trailing quotes and spaces
item_rows = item_rows.applymap(lambda x: x.strip(" '") if isinstance(x, str) else x)

# Extract item number and description
item_rows['item_no'] = item_rows['item_description'].apply(lambda x: x.split('.')[0].strip())
item_rows['item_description'] = item_rows['item_description'].apply(lambda x: '.'.join(x.split('.')[1:]).strip())


# Convert to structured JSON
parsed_data = item_rows[[
    'item_no', 'item_description', 'yes', 'sometimes', 'not_yet', 'score'
]].to_dict(orient='records')
```
